### Aim
To implement Monte Carlo (MC) Methods in Reinforcement Learning using Python to estimate state values $V(s)$ using First-Visit and Every-Visit Monte Carlo approaches on GridWorld environments, and to compare both methods using episode-based return calculations.

```
import numpy as np
import matplotlib.pyplot as plt
from monte_carlo.simulator import run_infographic_example, MonteCarloSimulation
```

### 1. Part 1: Infographic Example ($\gamma = 0.9$)
Calculating discounted returns $G_t$ across a 4-step trajectory.

```
info_returns = run_infographic_example(gamma=0.9)
print("State    Return G     New V(s)")
print("-" * 32)
for s, g in info_returns.items():
    print(f"{s:<8} {g:<12} V({s}) = {g}")
```

### 2. Part 2: GridWorld Simulations (2x3 and 3x3)
Comparing First-Visit MC and Every-Visit MC on multi-step trajectories with loops.

```
episodes_2x3 = [
    [("s4","U",-1), ("s1","R",-1), ("s2","L",-3), ("s1","stop",-1)],
    [("s3","U",-1), ("s0","stop",50)],
    [("s5","L",-4), ("s4","U",-2), ("s1","R",-1), ("s2","stop",-3)],
]
mc1 = MonteCarloSimulation(rows=2, cols=3, terminal_state="S0", gamma=1.0)
mc1.set_episodes(episodes_2x3)
V1_fv, _ = mc1.first_visit_mc()
V1_ev, _ = mc1.every_visit_mc()
print("2x3 First-Visit Values:", V1_fv)
print("2x3 Every-Visit Values:", V1_ev)
```

### 3. Conclusion & Findings
Monte Carlo methods successfully estimated state values directly from complete sample episodes. First-Visit MC provides unbiased estimates by considering only the initial encounter per episode, while Every-Visit MC leverages all visits, converging to identical expectations as the sample size grows.
